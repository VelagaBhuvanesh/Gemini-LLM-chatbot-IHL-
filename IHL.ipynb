```python
from google import genai
from google.genai import types
import numpy as np

def generate_embedding(text, task_type="RETRIEVAL_DOCUMENT"):

    response = client.models.embed_content(
        model="gemini-embedding-001",
        contents=text,
        config=types.EmbedContentConfig(
            task_type=task_type
        )
    )

    return response.embeddings[0].values
```

```python
from google.colab import files

uploaded = files.upload()

pdf_path = list(uploaded.keys())[0]

print("Uploaded:", pdf_path)
```

```python
from pypdf import PdfReader

reader = PdfReader(pdf_path)

document_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        document_text += text + "\n"

print("Text extracted successfully!")
print("Total characters:", len(document_text))
print(document_text[:1000])
```

```python
def chunk_text(text, chunk_size=500, overlap=100):

    words = text.split()
    chunks = []

    start = 0

    while start < len(words):

        end = start + chunk_size
        chunk = " ".join(words[start:end])

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks


chunks = chunk_text(document_text)

print("Total chunks:", len(chunks))
print("\nFirst chunk:\n", chunks[0])
```

```python
embeddings = []

for i, chunk in enumerate(chunks):
    embedding = generate_embedding(chunk)
    embeddings.append(embedding)

    if (i + 1) % 10 == 0:
        print(f"Processed {i + 1}/{len(chunks)} chunks")

import numpy as np

embeddings = np.array(embeddings, dtype="float32")

print("Embedding matrix shape:", embeddings.shape)
```

```python
import faiss
import numpy as np

# Normalize embeddings
faiss.normalize_L2(embeddings)

# Get embedding dimensions
dimension = embeddings.shape[1]

# Create FAISS index
index = faiss.IndexFlatIP(dimension)

# Add embeddings to the index
index.add(embeddings)

print("FAISS database created successfully!")
print("Total vectors stored:", index.ntotal)
```

```python
def retrieve_context(question, top_k=3):

    # Generate embedding for the question
    query_embedding = generate_embedding(
        question,
        task_type="RETRIEVAL_QUERY"
    )

    # Convert into NumPy array
    query_vector = np.array(
        [query_embedding],
        dtype="float32"
    )

    # Normalize query vector
    faiss.normalize_L2(query_vector)

    # Search FAISS database
    scores, indices = index.search(query_vector, top_k)

    # Retrieve matching chunks
    results = []

    for idx in indices[0]:
        if idx != -1:
            results.append(chunks[idx])

    return results
```

```python
question = input("Ask a question about your PDF: ")

results = retrieve_context(question)

for i, result in enumerate(results):

    print(f"\n--- Retrieved Chunk {i+1} ---")
    print(result)
```

```python
def retrieve_with_scores(question, top_k=3):

    query_embedding = generate_embedding(
        question,
        task_type="RETRIEVAL_QUERY"
    )

    query_vector = np.array(
        [query_embedding],
        dtype="float32"
    )

    faiss.normalize_L2(query_vector)

    scores, indices = index.search(query_vector, top_k)

    for rank, (score, idx) in enumerate(
        zip(scores[0], indices[0]), 1
    ):

        if idx != -1:
            print(f"\nRank: {rank}")
            print(f"Similarity: {score:.4f}")
            print(f"Chunk: {chunks[idx][:500]}")
```

```python
import time
from google.genai import errors

def data_engineering_chat(question):

    context_chunks = retrieve_context(question, top_k=3)

    context = "\n\n".join(context_chunks)

    prompt = f"""
{SYSTEM_PROMPT}

Use the following retrieved context to answer the question.

RETRIEVED CONTEXT:
{context}

USER QUESTION:
{question}

Provide a clear, accurate, and practical answer.
"""

    for attempt in range(3):

        try:
            response = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=prompt
            )

            return response.text

        except errors.ServerError as e:

            print(f"Server temporarily unavailable. Attempt {attempt + 1}/3")

            if attempt < 2:
                time.sleep(2 ** (attempt + 1))
            else:
                return "Gemini is currently experiencing high demand. Please try again later."

        except errors.ClientError as e:

            return f"API request failed: {e}"
```

```python
while True:

    question = input("\nYou: ")

    if question.lower() in ["exit", "quit", "stop"]:
        print("Chatbot: Goodbye!")
        break

    answer = data_engineering_chat(question)

    print("\nData Engineering AI:", answer)
```
